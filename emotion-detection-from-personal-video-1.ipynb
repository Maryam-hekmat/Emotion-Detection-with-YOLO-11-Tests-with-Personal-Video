```python
import os
from ultralytics import YOLO
```

```python
Root_dir = "/kaggle/input/8-facial-expressions-for-yolo/9 Facial Expressions you need"


yaml_path  = os.path.join(Root_dir, "data.yaml")
train_path = os.path.join(Root_dir, "train", "images")
valid_path = os.path.join(Root_dir, "valid", "images")
test_path  = os.path.join(Root_dir, "test", "images")

```

```python
assert os.path.exists(yaml_path), "data.yaml پیدا نشد!"
```

```python
def load_images(folder, max_images=None):
    images = []
    for f in os.listdir(folder):
        if f.lower().endswith(('.jpg', '.png', '.jpeg')):
            images.append(os.path.join(folder, f))
            if max_images and len(images) >= max_images:
                break
    return images
```

```python
train_images = load_images(train_path, max_images=100)
valid_images = load_images(valid_path, max_images=50)
test_images  = load_images(test_path,  max_images=50)

print("Train:", len(train_images))
print("Valid:", len(valid_images))
print("Test :", len(test_images))
```

```python
model = YOLO("yolo11n.pt")  # Make sure the model is on the Kaggle path

results = model.train(
    data=yaml_path,
    epochs=10,
    batch=16,     
    lr0=0.0001,
    lrf=0.1,
    imgsz=416,
    plots=True
)

print(results)
```

```python
from ultralytics import YOLO
import os

# مسیر ویدیوی شخصی
video_path = "/kaggle/input/datasets/maryamhekmat/my-face/WhatsApp Video 2025-12-15 at 17.34.02 (1).mp4"

# مدل آموزش دیده
model_path = "/kaggle/working/runs/detect/train/weights/best.pt"
model = YOLO(model_path)

# تنظیم confidence threshold (مثلاً 0.6)
model.conf = 0.6

# اجرای inference روی ویدیو و ذخیره خروجی
results = model(video_path, save=True)

# مسیر خروجی
output_path = results[0].save_dir
print(f"Video tested and saved at {output_path}")
```

```python
import cv2

input_path = "/kaggle/working/runs/detect/predict/WhatsApp Video 2025-12-15 at 17.34.02 (1).avi"
output_path = "/kaggle/working/output.mp4"

cap = cv2.VideoCapture(input_path)

# مشخصات ویدیو
fps = int(cap.get(cv2.CAP_PROP_FPS))
width  = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*"mp4v")
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

while True:
    ret, frame = cap.read()
    if not ret:
        break
    out.write(frame)

cap.release()
out.release()

print("Done.:", output_path)
```
